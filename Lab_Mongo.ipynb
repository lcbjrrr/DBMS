## Python to connect to Mongo

Python integrates with MongoDB primarily through the `pymongo` driver. This driver allows Python applications to connect to MongoDB databases, perform CRUD (Create, Read, Update, Delete) operations on collections and documents, and manage data programmatically. It translates Python objects (like dictionaries) into BSON (Binary JSON), which is MongoDB's native data format, and vice-versa, making data manipulation intuitive for Python developers.

````python
!pip install pymongo
````

````python
from pymongo import MongoClient
# Connect to the local MongoDB instance
client = MongoClient("mongodb://localhost:27017/")
# Equivalent to db.adminCommand({ listDatabases: 1 })
database_names = client.list_database_names()
print("List of databases:", database_names)
# Close the connection
client.close()
````

````python
!pip install pymongo
from pymongo import MongoClient

# Connect to the local MongoDB instance
client = MongoClient("mongodb://localhost:27017/")
db = client["local"]

# 1. db.createCollection("tab1")
# In pymongo, collection is created automatically on first insert if it doesn't exist.
# You can explicitly create it if needed, but it's often not necessary.
# db.create_collection("tab1")
print("Attempting to create collection 'tab1' (or use existing one).")
tab1 = db["tab1"]

# 2. db.tab1.insertOne({title: 'MongoDB Overview',description: 'MongoDB is no sql database'})
document1 = {"title": "MongoDB Overview", "description": "MongoDB is no sql database"}
insert_result = tab1.insert_one(document1)
print(f"Inserted document with id: {insert_result.inserted_id}")

# Add another document for better demonstration of find()
document2 = {"title": "Python with MongoDB", "description": "Connecting Python to MongoDB"}
tab1.insert_one(document2)

# 3. db.getCollectionNames()
collection_names = db.list_collection_names()
print("Collection names in 'mydatabase':", collection_names)

# 4. db.tab1.find()
print("\nAll documents in 'tab1':")
for doc in tab1.find():
    print(doc)

# Close the connection
client.close()
````
